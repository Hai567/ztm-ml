### Matplotlib example workflow

```python
# 0. import matplotlib and get ready
%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare data
x = [1,2,3,4]
y=[11,22,33,44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10,10)) # width, height

# 3. Plot data
ax.plot(x, y)

# 4. Customize plot
ax.set(title="Example plot", xlabel="x-axis", ylabel="y-axis")

# 5. Save & show
fig.savefig("./example_plot.png")
```

## 3. Different types of plot
- Line plot
- Scatter plot
- Bar plot
- Histogram
- Subplot

```python
# Making figures with NumPy arrays
x = np.linspace(0, 10, 100) # start, stop, number of data
x[:10]
```

### Line plot

```python
# Plot the data
fig, ax = plt.subplots()
ax.plot(x, x**4)
```

### Scatter plot

```python
fig, ax = plt.subplots()
ax.scatter(x, x**4)
fig.show()
```

```python
# Another scatter
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
```

### Bar plot

```python
prices = {"Almond": 10, "Peanut": 8, "Cashew": 12}
fig, ax = plt.subplots()
ax.set(title="Nuts' prices", ylabel="Price", xlabel="Nut")
ax.bar(x=prices.keys(), height=prices.values())
```

### Bar Horizontal

```python
fig, ax = plt.subplots()
ax.barh(y=prices.keys(), width=prices.values())
```

### Histogram

```python
x = np.random.randn(10000) # Random distribution
fig, ax = plt.subplots()
ax.hist(x)
```

### Subplots in a figure

```python
fig, ((ax1, ax2, ax3), (ax4, ax5, ax6)) = plt.subplots(nrows=2, ncols=3, figsize=(10, 5))

ax1.plot(x, x/2)
ax2.plot(x, x**2)
ax3.bar(x=prices.keys(), height=prices.values())
ax4.barh(y=prices.keys(), width=prices.values())
ax5.hist(x)
ax6.scatter(x, x**2)
```

```python
# 2nd way of subplot
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))
ax[0,0].scatter(x, x/2)
ax[1,1].hist(x)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../pandas/49 - car-sales.csv')

# Get the value counts for the 'Make' column
# .sort_values(ascending=False) ensures the bars are sorted by count
make_counts = df['Make'].value_counts().sort_values(ascending=False)

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(10, 6), squeeze=True) # Set a good figure size# Add grid lines for the y-axis

ax[0].pie(make_counts)
ax[1].bar(x=make_counts.index, height=make_counts.values)


```

### PyPlot vs matplotlib OO method comparison
- When plotting something quickly, it's ok to use PyPlot method
- When plotting advanced things, use the OO method

```python
over_50 = health_disease[health_disease['age'] > 50]
over_50[:5]
```

```python
over_50.plot(kind="scatter", x='age', y="chol", c="target", figsize=(10,6))
```

```python
fig, ax = plt.subplots()
over_50.plot(kind="scatter",
            x="age",
            y="chol",
            c="target",
            ax=ax)# difference is here
ax.set_xlim([0, 100])
```
