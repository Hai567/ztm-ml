### NumPy uses broadcasting to make everything faster
- Broadcasting: how numpy treats arrays with different shapes during arithmetic operations
- Broadcast the smaller array across the bigger one
- NumPy compares shapes element-wise
- 2 2D arrays are compatible when:
    - They are equal, or
    - One of them is 1
