- Broadcasting: how numpy treats arrays with different shapes during arithmetic operations
- Broadcast the smaller array across the bigger one
- NumPy compares shapes element-wise
- 2 2D arrays are compatible when:
    - They are equal, or
    - One of them is 1


### Aggregation

Aggregation = performing the same operation on a number of things

> use Python's methods (`sum()`) on Python's datatypes

> use NumPy's methods (`np.sum()`) on ndarrays

```python
massive_array = np.random.randint(0, 10, size = 100000)
massive_array[:5]
```

### Dot product & Element wise
- Dot product: A way to find a pattern between 2 sets of number

![.](./dot-product-vs-element-wise.png)
![.](./dot_product_colored.png)

```python
np.random.seed(2)

mat1 = np.random.randint(0, 10, size=(5,3))
mat2 = np.random.randint(0, 10, size=(3,5))

mat1, mat2
```

```python
# Element-wise multiplication (Hadamard Product)
np.multiply(mat1, mat2.T)
```

```python
# Dot Product
np.dot(mat1, mat2)
```
