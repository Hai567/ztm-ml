### I. A complete workflow

#### 1. Getting the data ready

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

np.random.seed(42)

heart_disease = pd.read_csv("./data/heart-disease.csv")
heart_disease.head(5)
```

```python
# accuracy metric
clf.score(X_train, Y_train)
```

```python
clf.score(X_test, Y_test)
```

```python
# other metrics
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print(classification_report(Y_test, Y_preds))
```

Option 2: Fill mising data with sklearn

```python
car_sales_missing = pd.read_csv("./data/car-sales-extended-missing-data.csv")
```

```python
# Drop rows with missing price 
car_sales_missing = car_sales_missing.dropna(subset="Price")
```

```python
# split data into training-testing sets before filling data
X = car_sales_missing.drop("Price", axis=1)
Y = car_sales_missing["Price"]

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
```

```python
# create imputers
from sklearn.impute import SimpleImputer
# Categorial imputer
cat_imputer = SimpleImputer(strategy="constant", fill_value="missing")
# Doors imputer
doors_imputer = SimpleImputer(strategy="constant", fill_value=4)
# Odometer imputer
num_imputer = SimpleImputer(strategy="mean")

# Define columns to apply imputers to
cat_features = ["Make", "Colour"]
doors_features = ["Doors"]
num_features = ["Odometer (KM)"]

```

```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer([
    ("cat_imputer", cat_imputer, cat_features),
    ("door_imputer", doors_imputer, doors_features),
    ("num_imputer", num_imputer, num_features)
], remainder="passthrough",
    sparse_threshold=0)

filled_X_train = pd.DataFrame(transformer.fit_transform(X_train), columns=["Make", "Colour", "Doors", "Odometer (KM)"])
filled_X_train.head(3)
```

```python
filled_X_test = pd.DataFrame(transformer.fit_transform(X_test), columns=["Make", "Colour", "Doors", "Odometer (KM)"])
filled_X_test.head(3)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

one_hot_features = ["Make", "Colour"]

one_hot = OneHotEncoder()
tranformer = ColumnTransformer([("one_hot", one_hot, one_hot_features)],
                               remainder="passthrough",
                               sparse_threshold=0)

transformed_X_train = tranformer.fit_transform(filled_X_train)
transformed_X_test = tranformer.fit_transform(filled_X_test)
```

```python
transformed_X_train
```

```python
np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(transformed_X_train, Y_train)

model.score(transformed_X_test, Y_test)
```

```python
# Get California Housing dataset
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
print(housing)
```

```python
housingdf = pd.DataFrame(housing["data"], columns=housing["feature_names"])
housingdf["Price"] = housing["target"]
housingdf.head()
```

```python
from sklearn.linear_model import Ridge
X = housingdf.drop("Price", axis=1)
Y = housingdf["Price"]

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
```

```python
model = Ridge()
model.fit(X_train, Y_train)

model.score(X_test, Y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(X_train, Y_train)

model.score(X_test, Y_test)
```
