## .loc & .iloc
- .loc: refer to index
- .iloc: refer to position

- both can use slicing [:2]

```python
# .loc (location) & .iloc
animals = pd.Series(['bird', 'dog', 'cat', 'frog',' horse', 'pig', 'ox'], index=[3,2,3,9,1,2,5])
```

```python
animals.loc[3]
```

```python
animals.loc[9]
```

```python
car_sales.loc[3]
```

```python
animals.iloc[3]
```

```python
animals.iloc[9]
```

```python
animals.iloc[:3]
```

```python
animals.loc[:3]
```

```python
car_sales.loc[:3]
```

```python
car_sales.Colour[:3]
```
