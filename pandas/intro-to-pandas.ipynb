2 main datatypes:
- series: array alike - 1-dimensional
- dataFrame: object alike - 2-dimensional

```python
# Series
series = pd.Series(["BMW", "Toyota", "Honda"])
colors = pd.Series(["blue", "red", "green"])
```

```python
car = pd.DataFrame({"Manufacturer": series, "Color": colors})
```
